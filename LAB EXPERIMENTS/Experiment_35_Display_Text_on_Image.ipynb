```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
```

```python
height = 500
width = 500

image = np.ones((height, width, 3), dtype=np.uint8) * 255

print("White image created successfully!")
```

```python
text = input("Enter the text to display on the image: ")

print("Entered text:", text)
```

```python
font = cv2.FONT_HERSHEY_SIMPLEX

position = (50, 250)
font_scale = 1
color = (0, 0, 255)  # Red color
thickness = 2

cv2.putText(
    image,
    text,
    position,
    font,
    font_scale,
    color,
    thickness,
    cv2.LINE_AA
)

print("Text displayed successfully!")
```

```python
plt.figure(figsize=(7, 6))

plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("User-Entered Text on Image")
plt.axis("off")

plt.show()
```

```python
output_name = "text_on_image.jpg"

cv2.imwrite(output_name, image)

files.download(output_name)

print("Output image saved successfully!")
```
