```python
import cv2
import pytesseract
import matplotlib.pyplot as plt
from google.colab import files
from IPython.display import display
```

```python
uploaded = files.upload()

video_path = next(iter(uploaded))

cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    raise ValueError("Unable to open video. Upload a valid video file.")

print("Video uploaded successfully.")
```

```python
extracted_text = []

frame_number = 0

while True:
    ret, frame = cap.read()

    if not ret:
        break

    # Process every 20th frame
    if frame_number % 20 == 0:

        gray = cv2.cvtColor(
            frame,
            cv2.COLOR_BGR2GRAY
        )

        text = pytesseract.image_to_string(gray)

        if text.strip():
            extracted_text.append(
                f"Frame {frame_number}:\n{text.strip()}"
            )

    frame_number += 1

cap.release()

print("Text extraction completed.")
```

```python
if extracted_text:
    for text in extracted_text:
        print(text)
        print("-" * 40)
else:
    print("No text detected in the video.")
```

```python
output_path = "extracted_video_text.txt"

with open(output_path, "w", encoding="utf-8") as file:
    file.write("\n\n".join(extracted_text))

print("Extracted text saved successfully.")
```
