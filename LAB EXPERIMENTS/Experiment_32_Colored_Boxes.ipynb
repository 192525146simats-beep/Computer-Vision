```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files
```

```python
height = 500
width = 500

image = np.ones((height, width, 3), dtype=np.uint8) * 255

print("White image created successfully!")
```

```python
box_width = width // 10
box_height = height // 10

# Black box
cv2.rectangle(image, (50, 50),
              (50 + box_width, 50 + box_height),
              (0, 0, 0), -1)

# Blue box
cv2.rectangle(image, (200, 50),
              (200 + box_width, 50 + box_height),
              (255, 0, 0), -1)

# Green box
cv2.rectangle(image, (50, 200),
              (50 + box_width, 200 + box_height),
              (0, 255, 0), -1)

# Red box
cv2.rectangle(image, (200, 200),
              (200 + box_width, 200 + box_height),
              (0, 0, 255), -1)

print("Four colored boxes created successfully!")
```

```python
plt.figure(figsize=(6, 6))
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("White Image with Four Colored Boxes")
plt.axis("off")
plt.show()
```

```python
output_name = "four_colored_boxes.jpg"

cv2.imwrite(output_name, image)

files.download(output_name)

print("Image saved successfully!")
```
