```python

import cv2
import matplotlib.pyplot as plt
from google.colab import files
from google.colab.patches import cv2_imshow

# Step 1: Upload an image dynamically
uploaded = files.upload()

# Step 2: Read the uploaded image
image_path = list(uploaded.keys())[0]
img = cv2.imread(image_path)

if img is None:
    raise ValueError("Unable to load image!")

# Step 3: Convert image to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Step 4: Load Haar Cascade for eye detection
eye_cascade = cv2.CascadeClassifier(
    cv2.data.haarcascades + 'haarcascade_eye.xml'
)

# Check classifier
if eye_cascade.empty():
    raise IOError("Eye cascade file not loaded!")

# Step 5: Detect eyes
eyes = eye_cascade.detectMultiScale(
    gray,
    scaleFactor=1.1,
    minNeighbors=5,
    minSize=(20, 20)
)

# Step 6: Draw rectangles around detected eyes
for (x, y, w, h) in eyes:
    cv2.rectangle(img, (x, y), (x+w, y+h), (0, 255, 0), 2)
    cv2.putText(img, "Eye", (x, y-10),
                cv2.FONT_HERSHEY_SIMPLEX, 0.6,
                (0, 255, 0), 2)

# Step 7: Display results
print("Number of eyes detected:", len(eyes))

plt.figure(figsize=(10, 7))
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Eye Detection Using OpenCV")
plt.axis("off")
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt
from google.colab import files

# Dynamic image upload
uploaded = files.upload()

# Read uploaded image
path = list(uploaded.keys())[0]
img = cv2.imread(path)

if img is None:
    raise ValueError("Invalid image!")

# Convert to grayscale
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Load eye cascade
cascade_path = cv2.data.haarcascades + "haarcascade_eye.xml"
eye_cascade = cv2.CascadeClassifier(cascade_path)

# Detect eyes
eyes = eye_cascade.detectMultiScale(
    gray,
    scaleFactor=1.1,
    minNeighbors=5,
    minSize=(20, 20)
)

# Draw rectangles
for (x, y, w, h) in eyes:
    cv2.rectangle(img, (x, y), (x+w, y+h), (0, 255, 0), 2)

# Display result
print("Number of eyes detected:", len(eyes))

plt.figure(figsize=(10, 7))
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
plt.title("Human Eye Detection")
plt.axis("off")
plt.show()
```
