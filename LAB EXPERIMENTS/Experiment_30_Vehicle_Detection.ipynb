```python
import cv2
import numpy as np
import os

cap = cv2.VideoCapture(video_name)

if not cap.isOpened():
    raise ValueError("Cannot open video.")

# Read video properties
fps = cap.get(cv2.CAP_PROP_FPS)

if fps <= 0:
    fps = 25

# Get dimensions from an actual frame
ret, first_frame = cap.read()

if not ret:
    raise ValueError("Cannot read video frames.")

height, width = first_frame.shape[:2]

# Vehicle classes
classes = [
    "background", "aeroplane", "bicycle", "bird", "boat",
    "bottle", "bus", "car", "cat", "chair", "cow",
    "diningtable", "dog", "horse", "motorbike",
    "person", "pottedplant", "sheep", "sofa", "train",
    "tvmonitor"
]

vehicle_ids = [6, 7, 14]

output_name = "/content/vehicle_detection_output.mp4"

out = cv2.VideoWriter(
    output_name,
    cv2.VideoWriter_fourcc(*"mp4v"),
    fps,
    (width, height)
)

if not out.isOpened():
    cap.release()
    raise RuntimeError("Unable to create output video.")

max_vehicles = 0

# Process the first frame and remaining frames
while ret:

    frame = first_frame.copy()

    blob = cv2.dnn.blobFromImage(
        cv2.resize(frame, (300, 300)),
        0.007843,
        (300, 300),
        127.5
    )

    net.setInput(blob)
    detections = net.forward()

    count = 0

    for i in range(detections.shape[2]):

        confidence = detections[0, 0, i, 2]
        class_id = int(detections[0, 0, i, 1])

        if confidence > 0.50 and class_id in vehicle_ids:

            box = detections[0, 0, i, 3:7] * np.array(
                [width, height, width, height]
            )

            x1, y1, x2, y2 = box.astype(int)

            x1 = max(0, x1)
            y1 = max(0, y1)
            x2 = min(width - 1, x2)
            y2 = min(height - 1, y2)

            if x2 <= x1 or y2 <= y1:
                continue

            cv2.rectangle(
                frame, (x1, y1), (x2, y2),
                (0, 255, 0), 2
            )

            cv2.putText(
                frame,
                classes[class_id],
                (x1, max(y1 - 10, 20)),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.6,
                (0, 255, 0),
                2
            )

            count += 1

    max_vehicles = max(max_vehicles, count)

    out.write(frame)

    ret, first_frame = cap.read()

cap.release()
out.release()

print("Vehicle detection completed!")
print("Maximum vehicles in one frame:", max_vehicles)
print("Output file created:", os.path.exists(output_name))
```
