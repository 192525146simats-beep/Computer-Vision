```python
import cv2
from google.colab import files
from IPython.display import Video, display
```

```python
uploaded = files.upload()

video_path = next(iter(uploaded))

cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    raise ValueError("Unable to open video. Upload a valid video file.")

fps = cap.get(cv2.CAP_PROP_FPS)

print("Original FPS:", fps)
```

```python
frames = []

while True:
    ret, frame = cap.read()

    if not ret:
        break

    frames.append(frame)

cap.release()

print("Total frames:", len(frames))
```

```python
if len(frames) == 0:
    raise ValueError("No frames found in the video.")

reversed_frames = frames[::-1]

print("Video frames reversed successfully.")
```

```python
height, width = reversed_frames[0].shape[:2]

output_path = "reverse_slow_motion.mp4"

# Reduce playback speed to half
slow_fps = max(fps / 2, 1)

fourcc = cv2.VideoWriter_fourcc(*'mp4v')

out = cv2.VideoWriter(
    output_path,
    fourcc,
    slow_fps,
    (width, height)
)

for frame in reversed_frames:
    out.write(frame)

out.release()

print("Reversed slow-motion video saved successfully.")
```

```python
display(Video(output_path, embed=True))
```
