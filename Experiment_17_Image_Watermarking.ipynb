```python

from google.colab import files
uploaded = files.upload()
```

```python

import cv2
import matplotlib.pyplot as plt

filename = list(uploaded.keys())[0]
image = cv2.imread(filename)

print("Image uploaded successfully:", filename)
```

```python

watermark = image.copy()

cv2.putText(
    watermark, "SIMATS", (50, 50),
        cv2.FONT_HERSHEY_SIMPLEX, 1,
            (0, 0, 255), 2
            )

plt.imshow(cv2.cvtColor(watermark, cv2.COLOR_BGR2RGB))
plt.title("Image with Text Watermark")
plt.axis("off")
plt.show()
```

```python

overlay = image.copy()

cv2.putText(
    overlay, "SIMATS", (50, 100),
        cv2.FONT_HERSHEY_SIMPLEX, 2,
            (0, 0, 255), 3
            )

blended = cv2.addWeighted(image, 0.7, overlay, 0.3, 0)

plt.imshow(cv2.cvtColor(blended, cv2.COLOR_BGR2RGB))
plt.title("Transparent Watermark")
plt.axis("off")
plt.show()
```
